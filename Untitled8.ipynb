```python
import os
from pathlib import Path
import hashlib

# Function to hash files and remove duplicates
def file_hash(filepath):
    with open(filepath, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

def remove_duplicates(folder):
    seen = {}
    removed = []
    for root, _, files in os.walk(folder):
        for file in files:
            path = Path(root) / file
            h = file_hash(path)
            if h in seen:
                os.remove(path)
                removed.append(path)
            else:
                seen[h] = path
    return removed

# Function to remove unwanted file types
valid_ext = {".jpg", ".jpeg", ".png"}

def clean_extensions(folder):
    removed = []
    for root, _, files in os.walk(folder):
        for file in files:
            path = Path(root) / file
            if path.suffix.lower() not in valid_ext:
                os.remove(path)
                removed.append(path)
    return removed
```

```python
dataset_path = "/content/drive/MyDrive/FASDD_PROJECT"

removed_files = remove_duplicates(dataset_path)
print(f"Removed {len(removed_files)} duplicates")

removed_ext = clean_extensions(dataset_path)
print(f"Removed {len(removed_ext)} unwanted files")
```

```python
import cv2

def is_blurry(image_path, threshold=100):
    img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return False
    variance = cv2.Laplacian(img, cv2.CV_64F).var()
    return variance < threshold

blurry_removed = []
for root, _, files in os.walk(dataset_path):
    for file in files:
        path = Path(root) / file
        if is_blurry(path):
            os.remove(path)
            blurry_removed.append(path)

print(f"Removed {len(blurry_removed)} blurry images")
```

```python
from PIL import Image

def is_truncated(path):
    try:
        with Image.open(path) as im:
            im.verify()
        return False
    except Exception:
        return True
```
